# Expectaion Maximization Algortihm

As discussed in Network Science course at Oklahoma University



```python
import math
import numpy as np
import matplotlib.pyplot as plt

def likelihood(x,mu,sigma):
  return 1/math.sqrt(2*math.pi*sigma**2)*math.e**((x-mu)**2/(2*sigma**2))

def expectation(xlist,mu,sigma):
  s = sum([likelihood(x,mu,sigma) for x in xlist])
  return [likelihood(x,mu,sigma)/s for x in xlist]

def newmu(xlist,mu,sigma):
  s = sum([likelihood(xlist[i],mu,sigma) for i in range(len(xlist))])
  return sum([xlist[i]*expectation(xlist,mu,sigma)[i] for i in range(len(xlist))])/s

x = [1,2,4,5]

mu = sum(x)/len(x)
sigma = sum([(x[i]-mu)**2 for i in range(len(x))])/len(x)
```

```python
while abs(mu - newmu(x,mu,sigma))>0.0001:
  mu = newmu(x,mu,sigma)

print(mu)
```
